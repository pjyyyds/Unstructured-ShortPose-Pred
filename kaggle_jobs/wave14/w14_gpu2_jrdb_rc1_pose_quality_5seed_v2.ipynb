```python
import hashlib, os, subprocess, sys, tempfile
from pathlib import Path
import torch

WORK = Path('/kaggle/working')
assert WORK.is_dir() and torch.cuda.is_available(), 'Kaggle GPU required'
REV = '1f1b81177290c8ed282d40723ac3c7d7f0a34743'
with tempfile.TemporaryDirectory(prefix='w14_jrdb_v2_') as ts:
    tmp = Path(ts)
    repo = tmp / 'repo'
    subprocess.run(['git', 'init', '-q', str(repo)], check=True)
    subprocess.run(['git', '-C', str(repo), 'fetch', '-q', '--depth=1', 'https://github.com/pjyyyds/Unstructured-ShortPose-Pred.git', REV], check=True)
    subprocess.run(['git', '-C', str(repo), 'checkout', '-q', '--detach', 'FETCH_HEAD'], check=True)
    env = {**os.environ, 'PYTHONPATH': str(repo), 'PYTHONDONTWRITEBYTECODE': '1'}
    dest = WORK / 'jrdb_rc1_pose_quality_and_initiation_5seed.zip'
    subprocess.run([sys.executable, '-u', '-m', 'scripts.run_jrdb_rc1_pose_quality_and_initiation_5seed', '--repo-root', str(repo), '--output-zip', str(dest)], cwd=repo, env=env, check=True)
    print('DONE:', dest, dest.stat().st_size, hashlib.sha256(dest.read_bytes()).hexdigest(), flush=True)
```
