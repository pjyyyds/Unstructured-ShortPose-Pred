# w19_gpu1_dut_imptc_and_unified_clean_chain
Inputs: `shortpose-wave1-private-inputs` + notebook output `w1-gpu2-dut-pose-and-graph` | Accelerator: `GPU T4 x2`


```python
import hashlib, os, subprocess, sys, tempfile, zipfile
from pathlib import Path
import torch

WORK, MOUNT = Path('/kaggle/working'), Path('/kaggle/input')
assert WORK.is_dir() and torch.cuda.is_available(), 'Kaggle GPU required'
n_gpu = torch.cuda.device_count()
REV = '1768f2b26b3a89dc3b799c109a418ae0ca3bdd43'
with tempfile.TemporaryDirectory(prefix='w19_g1_') as ts:
    tmp = Path(ts)
    repo, meta_dir = tmp / 'repo', tmp / 'pinns_meta'
    subprocess.run(['git', 'init', '-q', str(repo)], check=True)
    subprocess.run(['git', '-C', str(repo), 'fetch', '-q', '--depth=1', 'https://github.com/pjyyyds/Unstructured-ShortPose-Pred.git', REV], check=True)
    subprocess.run(['git', '-C', str(repo), 'checkout', '-q', '--detach', 'FETCH_HEAD'], check=True)
    base_env = {**os.environ, 'PYTHONPATH': str(repo), 'PYTHONDONTWRITEBYTECODE': '1'}
    joint_bin = next(MOUNT.rglob('joint_actions_all.bin'))
    heldout_bin = next(MOUNT.rglob('shortpose_heldout.bin'))
    map_bin = next(MOUNT.rglob('pinns_source_label_causal_maps.bin'))
    pose_npz = tmp / 'pose.npz'
    with zipfile.ZipFile(heldout_bin) as zf:
        pose_npz.write_bytes(zf.read('pose.npz'))
    z_hits = list(MOUNT.rglob('dut_shared_space_graph_transfer.zip'))
    if z_hits:
        dut_zip = z_hits[0]
    else:
        pg_f = next(MOUNT.rglob('private_graph.npz'))
        et_f = next(MOUNT.rglob('errors_and_targets.npz'))
        dut_zip = tmp / 'dut_graph.zip'
        with zipfile.ZipFile(dut_zip, 'w', zipfile.ZIP_DEFLATED) as dz:
            dz.write(pg_f, 'private_graph.npz')
            dz.write(et_f, 'errors_and_targets.npz')
    subprocess.run([sys.executable, '-m', 'scripts.stage_pinns_metadata', '--output-dir', str(meta_dir)], cwd=repo, env=base_env, check=True)
    d0 = WORK / 'wave19_dut_imptc_conflict_and_conformal_5seed.zip'
    d1 = WORK / 'wave19_unified_clean_m0_to_m3_5seed.zip'
    imptc_tar = tmp / 'imptc_samples.tar.gz'
    e0 = {**base_env, 'CUDA_VISIBLE_DEVICES': '0'}
    e1 = {**base_env, 'CUDA_VISIBLE_DEVICES': '1' if n_gpu > 1 else '0'}
    p0 = subprocess.Popen([
        sys.executable, '-u', '-m', 'scripts.run_wave19_dut_imptc_conflict_and_conformal_5seed',
        '--meta-dir', str(meta_dir), '--pose-npz', str(pose_npz), '--joint-zip', str(joint_bin),
        '--causal-map-bin', str(map_bin), '--dut-zip', str(dut_zip), '--imptc-tar', str(imptc_tar),
        '--zip-path', str(d0)
    ], cwd=repo, env=e0)
    p1 = subprocess.Popen([
        sys.executable, '-u', '-m', 'scripts.run_wave19_unified_clean_m0_to_m3_5seed',
        '--repo-root', str(repo), '--meta-dir', str(meta_dir), '--pose-npz', str(pose_npz),
        '--joint-zip', str(joint_bin), '--causal-map-bin', str(map_bin), '--zip-path', str(d1)
    ], cwd=repo, env=e1)
    rc0, rc1 = p0.wait(), p1.wait()
    assert rc0 == 0 and rc1 == 0, f'Subprocess failed: rc0={rc0}, rc1={rc1}'
    for d in (d0, d1):
        print('DONE:', d, d.stat().st_size, hashlib.sha256(d.read_bytes()).hexdigest(), flush=True)
```
